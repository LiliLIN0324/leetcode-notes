## 01 时间复杂度：算法的执行效率；算法的执行时间和算法的输入值的关系。
O(N) -> 在for i in range(num) 循环中，和O(num)有关
O(q) -> 和num无关

```python
# 时间复杂度
def o1(num):
    if num < 0:
        return False
    while num > 1:
        if num % 2 != 0:
            return False
        num = num // 2
    return True
# -> O(1)的原因是循环的次数是固定的，不随输入规模的变化而变化，所以时间复杂度是O(1)

def OlogN(num):
    i = 1
    while i < num:
        i *= 2  
    return i == num
# -> O(logN)的原因是i = i*2，循环的次数是log2(num)，所以时间复杂度是O(logN)
# 2^i = num => i = log2(num)

def ON(num):
    total = 0
    for i in range(num):
        total += i
    return total
# -> O(N)的原因是循环的次数是num，所以时间复杂度是O(N)

def OMN(num):
    total = 0
    for i in range(num):
        total += i
    for j in range(num):
        total += j
    return total
# -> O(M+N)的原因是外层循环是O(M)，内层循环是O(N)，所以总的时间复杂度是O(M+N)

def ONLogN(num1, num2):
    total = 0
    j = 0
    for i in range(num1):
        for j in range(num2):
            total += i + j
            j *= 2
    return total
# -> O(NlogN)的原因是外层循环是O(N)，内层循环是O(logN)，所以总的时间复杂度是O(NlogN)

def ON2(num):
    total = 0
    for i in range(num):
        for j in range(num):
            total += i + j
    return total
# -> O(N^2)的原因是外层循环是O(N)，内层循环是O(N)，所以总的时间复杂度是O(N^2)

# 时间复杂度：O(1)< O(logN)< O(N)< O(M+N)< O(NlogN)< O(N^2)

# -> 方法论： O(logN) -> 二分排序法
# -> 方法论： O(NlogN) -> 归并排序法
# -> 方法论： O(N^2) -> 暴力解法
```

# 27. 移除元素

```python
# 27. 移除元素
nums = [3,2,2,3] 
val = 3
nums = [0,1,2,2,3,0,4,2]
val = 2

# 方法一：快慢指针法
# 时间复杂度：O(n)，空间复杂度：O(1)
l = 0
for num in nums:
    if num != val:
        nums[l] = num
        l += 1
print(l)

# 方法二：双指针法
# 时间复杂度：O(n)，空间复杂度：O(1)
i = 0
j = len(nums) - 1
while i <= j:
    if nums[i] == val:
        nums[i] = nums[j]
        j -= 1
    else:
        i += 1
print(j-1)
```

# 203. 移除链表元素



```python
# 203. Remove Linked List Elements 移除链表元素

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
class Solution:
    def removeElements(self, head: Optional[ListNode], val: int) -> Optional[ListNode]:
        # 创建一个虚拟头节点，指向链表的头节点
        dummy = ListNode(0)
        dummy.next = head
        # 初始化当前节点为虚拟头节点
        current = dummy
        # 遍历链表
        while current.next:
            if current.next.val == val:
                # 如果当前节点的下一个节点的值等于val，则删除该节点
                current.next = current.next.next
            else:
                # 否则，移动到下一个节点
                current = current.next
        # 返回虚拟头节点的下一个节点，即新的链表头节点
        return dummy.next
        
```

# 215. 数组中第K个最大的元素

```python
# 暴力方法1
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        nums.sort(reverse=True)
        return(nums[k-1])

# 很明显这个解法的复杂度是高的 
# 时间复杂度：O(nlogn)，空间复杂度：O(1)
```

```python
# 最小堆解法
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        import heapq
        heap = []
        for num in nums:
            heapq.heappush(heap, num)
            print('heap:', heap)
            if len(heap) > k:
                heapq.heappop(heap)
                print('heap after pop:', heap)
        return heap[0]
    
nums = [3,2,1,5,6,4]
k = 2
solution = Solution()
print(solution.findKthLargest(nums, k))  # 输出: 5 

# 时间复杂度：O(nlogk)，空间复杂度：O(k)
```

```python
# 最大堆解法
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        import heapq
        heap = []
        # for num in nums:
        #     heapq.heappush(heap, -num)
        heap = [-num for num in nums]
        heapq.heapify(heap)
        # 拿走前 k-1 大的
        for _ in range(k - 1):
            heapq.heappop(heap)
        return -heap[0]
    
nums = [3,2,1,5,6,4]
k = 4
solution = Solution()
print(solution.findKthLargest(nums, k))  # 输出: 5 
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    # 定义小于比较规则：按分数比较
    def __lt__(self, other):
        if self.score == other.score:
            return self.name < other.name  # 如果分数相同，按名字比较
        return self.score < other.score

# class的调用方式是
s1 = Student("Alice", 90)
s2 = Student("Bob", 85)
s3 = Student("Bob", 75)
s4 = Student("Mike", 90)
# 使用 < 运算符，会自动调用 s1.__lt__(s2)
print(s1 < s2)  # 输出: False (因为 90 > 85)
```

```python
for a in [s1, s2, s3, s4]:
    print(a.name, a.score)
```
