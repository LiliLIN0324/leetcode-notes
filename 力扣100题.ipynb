# 1. 两数之和
给定一个整数数组 nums 和一个整数目标值 target，请你在该数组中找出 和为目标值 target  的那 两个 整数，并返回它们的数组下标。

你可以假设每种输入只会对应一个答案，并且你不能使用两次相同的元素。

你可以按任意顺序返回答案。

 
示例 1：

输入：nums = [2,7,11,15], target = 9
输出：[0,1]
解释：因为 nums[0] + nums[1] == 9 ，返回 [0, 1] 。
示例 2：

输入：nums = [3,2,4], target = 6
输出：[1,2]
示例 3：

输入：nums = [3,3], target = 6
输出：[0,1]
 

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        # 方法1: 暴力法
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]

nums = [2, 7, 11, 15]
target = 9

solution = Solution()
result = solution.twoSum(nums, target)
print(result)
```

# 128.最长连续序列

```python
class Solution(object):
    def longestConsecutive(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        # 中心思想：只从头判断，不是头的数字跳过
        st = set(nums)
        print(nums)
        print(st)
        # 换成集合set的原因是，集合是无序的，
        # 元素必须是唯一的，会自动去除重复数据。
        # 集合的查找时间复杂度是O(1)，而列表的查找时间复杂度是O(n)，所以用集合可以提高效率
        ans = 0
        m = len(st)
        for x in st:
            if x - 1 in st:
                continue # continue的意思是跳过当前循环，直接进入下一次循环
            # 这个写法的意思是，如果x-1在集合st中，说明x不是一个连续序列的起点，直接跳过
            length = 1
            nxt_x = x + 1
            while nxt_x in st:
            # 用while不用if的原因是，while可以一直循环，直到条件不满足为止，而if只会判断一次
                length += 1
                nxt_x += 1
            ans = max(ans, length)
            if ans *2 > m:
                break # break的意思是跳出当前循环，直接进入下一条语句
            # 这个写法的意思是，如果当前最长连续序列的长度已经大于集合st的一半，说明已经找到了最长的连续序列，可以直接跳出循环，提高效率
        return ans

nums = [100, 4, 200, 1, 3, 2]
solution = Solution()
result = solution.longestConsecutive(nums)
print(result)

# 这个方法总共时间复杂度是O(n)，空间复杂度是O(n)，其中n是数组nums的长度。
# 我使用了集合set,花费O(n)的时间和空间来存储数组中的元素，然后遍历集合中的每个元素，判断它是否是一个连续序列的起点，如果是，就继续向后查找连续的元素，直到找不到为止。
# 然后使用了for 循环和while循环来实现这个过程，最终返回最长连续序列的长度。分别时间复杂度是O(n)和O(n)，所以总时间复杂度是O(n)，空间复杂度是O(n)。
```

# 283. 移动零

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # 方法1: 双指针
        n = len(nums)
        left = 0 # left指针指向当前要放置非零元素的位置
        for right in range(n):  # right指针遍历整个数组
            if nums[right] != 0: # 如果right指针指向的元素不是0
                nums[left], nums[right] = nums[right], nums[left]
                left += 1

nums = [0, 1, 0, 3, 12]
solution = Solution()
result = solution.moveZeroes(nums)
print(nums)
```

# 11.盛最多水的容器

```python
import math
class Solution:
    def maxArea(self, height: List[int]) -> int:
        """
        :type height: List[int]
        :rtype: int
        """
        # 方法1: 双指针
        n = len(height)
        left, right = 0,n-1
        ans = -math.inf
        while left < right:
            ans = max(ans, (right - left) * min(height[left], height[right]))
            if height[right] <= height[left]:
                right -= 1
            else:
                left += 1
        return ans

input_height = [1,8,6,2,5,4,8,3,7]
solution = Solution()
result = solution.maxArea(input_height)
print(result)
```

# 15.三数之和

```python
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        """
        :type nums: List[int]
        :rtype: List[List[int]]
        """
        # 方法1: 排序 + 双指针
        nums_sorted=sorted(nums)
        ans = []
        n = len(nums_sorted)
        for i in range(n-2):
            if nums_sorted[i] > 0:
                break
            if i !=0 and nums_sorted[i] == nums_sorted[i-1]:
                continue 
            j = i + 1
            k = n - 1
            while j < k:
                s = nums_sorted[i] + nums_sorted[j] + nums_sorted[k]
                if s == 0:
                    ans.append([nums_sorted[i], nums_sorted[j], nums_sorted[k]])
                    j += 1
                    k -= 1
                    while j < k and nums_sorted[j] == nums_sorted[j + 1]:
                        j += 1
                    while j < k and nums_sorted[k] == nums_sorted[k - 1]:
                        k -= 1
                elif s < 0:
                    j += 1
                else:
                    k -= 1
        return ans

input_nums = [-1, 0, 1, 2, -1, -4]
solution = Solution()
result = solution.threeSum(input_nums)
print(result)
```

## 438.找到字符串中所有字母异位词

```python
# 方法1: 哈希表
# 复杂度分析
# 时间复杂度: O(n * m * log(m))，其中 n 是字符串 s 的长度，m 是每个子串的长度
# 空间复杂度: O(n * m)

from collections import defaultdict
class Solution:
    def findAnagrams(self, s: List[str], p: List[str]) -> List[List[str]]:
        """
        :type s: List[str]
        :type p: List[str]
        :rtype: List[List[str]]
        """

        mp = defaultdict(list)

        m = len(p)

        # p 排序后的结果
        p_sorted = ''.join(sorted(p))

        # 每次取 s 中长度为 m 的子串
        for i in range(len(s) - m + 1):

            snt = s[i:i+m]

            # 把子串排序
            snt_sorted = ''.join(sorted(snt))

            # 如果和 p 排序后一样，说明是异位词
            if snt_sorted == p_sorted:
                mp[p_sorted].append(i)

        return mp[p_sorted]

s='cbaebabacd'
p = 'abc'
solution = Solution()
result = solution.findAnagrams(s, p)
print(result)
```

## 3.******** 无重复字符的最长字串

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        # 时间复杂度: O(n)，空间复杂度: O(128)或(1)，因为字符集是有限的，最多只有128个ASCII字符
        ans = 0
        cnt = Counter()
        left = 0
        for right, c in enumerate(s):
            cnt[c] += 1
            while cnt[c] > 1:
                cnt[s[left]] -= 1
                left += 1
            ans = max(ans, right - left + 1)

        return ans

input_str = "abcabcbb"
solution = Solution()
result = solution.lengthOfLongestSubstring(input_str) 
print(result)  
```

# 239 ***[不懂].滑动窗口最大值

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        """
        :type nums: List[int]
        :type k: int
        :rtype: List[int]
        """
        # 方法0：暴力法
        # 时间复杂度: O(n*k)，空间复杂度: O(1)
        # n = len(nums)
        # ans = []
        # for i in range(n - k + 1):
        #     ans.append(max(nums[i:i + k]))
        # return ans

        # 方法1：单调队列
        # 时间复杂度: O(n)，空间复杂度: O(k)
        n = len(nums)
        ans = []
        q = deque() # 双端队列，存储当前窗口的最大值的索引
        for i in range(n):
            # 移除队列中不在窗口范围内的元素
            while q and q[0] < i - k + 1:
                q.popleft()
            # 移除队列中小于当前元素的所有元素，因为它们不可能成为最大值
            while q and nums[q[-1]] < nums[i]:
                q.pop()
            q.append(i) # 将当前元素的索引加入队列
            if i >= k - 1:
                ans.append(nums[q[0]]) # 当前窗口的最大值是队列的第一个元素
        return ans

nums = [1,3,-1,-3,5,3,6,7]
k = 3
solution = Solution()
result = solution.maxSlidingWindow(nums, k)
print(result)
```
