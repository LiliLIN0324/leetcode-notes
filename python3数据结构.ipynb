# 08 LinkedList 链表

由一系列的节点组成，包含数据域与指针域。

- 节点：`[数据域|指针域] = [val|next]`
- 链表：`[val|next] -> [val|next] -> [val|next] -> Null`

## 自定义类和对象，通过类表示节点。

数组和链表的区别是什么？

- 数组是连续的内存空间
- 而链表可以散在数据结构里面

```
[ , , , ,]
[ ,1, ,2,]
[4, ,3, ,]
```

- 链表：`[1|next] -> [2|next] -> [3|next] -> [4|next] -> Null`

- 数组优点：access 快；缺点：search、insert、delete 慢
- 链表优点：insert、delete 快；缺点：access、search 慢

- 访问 access -> O(N)
- 搜索 search -> O(N)
- 插入 insert -> O(1)
- **已知目标位置/节点的前驱时，修改链表指针是 O(1)。**
- **因为单链表没有 prev**
- 删除 delete -> O(1)

- A: 通常需要知道前驱节点才能修改其 next

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
# 封装一个方法帮助添加节点
def insert_node(head, val):
    new_node = ListNode(val)
    if head is None:
        return new_node
    else:
        current = head
        while current.next is not None:
            current = current.next
        current.next = new_node
        return head
    
if __name__ == "__main__":
    # 创建一个链表 1
    head = ListNode(1)
    head.next = ListNode(123)
    head.next.next = ListNode(456)
    head.next.next.next = ListNode(789)
    print()
    tmp = head
    tmp = tmp.next
    tmp.next = ListNode(100) # 增加100的节点
    head.next.next = None #删除100的节点
````

````python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
# val 和 next 会定义好

# 链表的常用操作
# 1.创建链表
# 2.添加元素
# 3.访问元素
# 4.查找元素
# 5.删除元素
# 6.链表的长度

# 1. 创建链表
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
# 2. 添加元素
new_node = ListNode(4)
new_node.next = head.next
head.next = new_node
# 3. 访问元素
current = head
while current:
    print(current.val)
    current = current.next
# 4. 查找元素
current = head
while current:
    if current.val == 2:
        print("Found:", current.val)
        break
    current = current.next
# 5. 删除元素
current = head
while current and current.next:
    if current.next.val == 3:
        current.next = current.next.next
        break
    current = current.next
# 6. 链表的长度
length = 0
current = head
while current:
    length += 1
    current = current.next
print("Length of linked list:", length) 
````

# 09 Hash Table 哈希表

- python 里就是 dictionary
- 格式 `key : value`
- 例子：`{'name':'AA', 'age':20}`


````python
# 哈希表常用的语法
hash_map = {}
hash_map['key'] = 'value'
print(hash_map['key'])  # 输出: value

# 检查键是否存在
if 'key' in hash_map:
    print('key exists')

# 获取键的值，如果不存在则返回默认值
print(hash_map.get('key', 'default'))  # 输出: value
print(hash_map.get('nonexistent', 'default'))  # 输出: default

values = hash_map.values()  # 获取所有值
print(values)  # 输出: dict_values(['value'])

# 输入value,输出对应的key
for k, v in hash_map.items():
    if v == 'value':
        print(k)  # 输出: key
        
````

**查找 O(1)**
因为可以把key 映射到存储位置

## 哈希冲突

发现：

     ↓
    桶3
     ↑

因此平均 O(1)

````python
# 12 heapq 堆
import heapq
from heapq import heappush, heappop, heapify, nsmallest, nlargest

# 小堆
a=[3, 1, 4, 1, 5, 9, 2, 6, 5]
# heapq.heapify(a)  # 将列表a转换为最小堆
print(a)  # [1, 1, 2, 3, 5, 9, 4, 6, 5]
heapq.heappush(a, 0)  # 向堆中添加元素0
print(a)  # [0, 1, 1, 3, 5, 9, 4, 6, 5, 2]
heapq.heappop(a)  # 从堆中删除最小元素
print(a)  # [1, 1, 2, 3, 5, 9, 4, 6, 5]

# 获取堆中最小的n个元素
print(nsmallest(3, a))  # [1, 1, 2]

# 获取堆中最大的n个元素
print(nlargest(3, a))  # [9, 6, 5]

# 大堆
a = [3, 1, 4, 1, 5, 9, 2, 6, 5]
b = [-x for x in a]  # 将列表a中的元素取反
heapq.heapify(b)  # 将列表b转换为最小堆
print(b)  # [-9, -6, -5, -5, -4, -3, -2, -1, -1]

# 从大堆中获取最大的n个元素
print([-x for x in nsmallest(3, b)])  # [9, 6, 5]
# 从大堆中获取最小的n个元素
print([-x for x in nlargest(3, b)])  # [5, 4, 3]
# 最后的大堆结果
print([-x for x in b])  # [9, 6, 5, 5, 4, 3, 2, 1, 1]
````

# 附1： 时间复杂度

````python
# 实现三个函数，时间复杂度分别是 $O(n)$ , $O(n^2)$ , $O(n\log n)$ ，使用加法运算来统一测试。

def linear_time_complexity(arr):
    result = 0
    for i in arr:
        result += i
    return result

def quadratic_time_complexity(arr):
    result = 0
    for i in arr:
        for j in arr:
            result += i * j
    return result
def triple_nested_time_complexity(arr):
    result = 0
    for i in arr:
        for j in arr:
            for k in arr:
                result += i * j * k
    return result
def nlogn_time_complexity(arr):
    result = 0
    for i in arr:
        result += i * (len(arr) // 2)
    return result
def logarithmic_time_complexity(arr):
    result = 0
    for i in arr:
        x = i

        while x > 1:
            x //= 2
            result += 1
    return result

if __name__ == "__main__":
    # 运行时间对比
    import time
    import random
    
    # 创建一个包含1000个元素的列表
    arr = [random.randint(1, 1000) for _ in range(1000)]

    # 测试线性时间复杂度
    start_time = time.time()
    result = linear_time_complexity(arr)
    end_time = time.time()
    print(f"Linear Time Complexity Result: {result}, Time: {end_time - start_time:.6f} seconds")

    # 测试二次时间复杂度
    start_time = time.time()
    result = quadratic_time_complexity(arr)
    end_time = time.time()
    print(f"Quadratic Time Complexity Result: {result}, Time: {end_time - start_time:.6f} seconds")

    # 测试对数时间复杂度
    start_time = time.time()
    result = logarithmic_time_complexity(arr)
    end_time = time.time()
    print(f"Logarithmic Time Complexity Result: {result}, Time: {end_time - start_time:.6f} seconds")

    # 测试nlogn时间复杂度
    start_time = time.time()
    result = nlogn_time_complexity(arr)
    end_time = time.time()
    print(f"NLogN Time Complexity Result: {result}, Time: {end_time - start_time:.6f} seconds")

    # 测试三次时间复杂度
    start_time = time.time()
    result = triple_nested_time_complexity(arr)
    end_time = time.time()
    print(f"Triple Nested Time Complexity Result: {result}, Time: {end_time - start_time:.6f} seconds")
````

## x^n 算法

````python
# x^n 算法
# 方法1：暴力法，复杂度是O(n)
def xx(x,n):
    result = 1
    for i in range(n):
        result = result *x
    return result
print(xx(2,10)) # 1024
````

````python
def xx2(x,n):
    return x**n # 复杂度是O(logn)
print(xx2(2,10)) # 1024
# 复杂度的原因是
````

````python
# 方法2：递归法，复杂度是O(logn)
def xx_recursive(x, n):
    if n == 0:
        return 1
    elif n % 2 == 0:
        half = xx_recursive(x, n / 2)
        return half * half
    else:
        return x * xx_recursive(x, n - 1)

print(xx_recursive(2, 10)) # 1024
# 解题思路是：
# 1. 如果n是偶数，那么x^n = (x^(n/2))^2
# 2. 如果n是奇数，那么x^n = x * x^(n-1)
````

````python
class Solution:
    def myPow(self, x: float, n: int) -> float:
        if n < 0:
            x = 1 / x
            n = -n

        if n == 0:
            return 1

        ans = self.myPow(x, n // 2) ##重要

        if n % 2 == 0:
            return ans * ans
        else:
            return ans * ans * x
        
s = Solution()
print(s.myPow(2, 10)) # 1024
````
