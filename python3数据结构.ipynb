# 08 LinkedList 链表
由一系列的节点组成，包含数据域与指针域。
[数据域|指针域] = [val|next] 
[val|next] -> [val|next] -> [val|next] -> Null

## 自定义类和对象，通过类表示节点。

数组和链表的区别是什么？
数组是连续的内存空间
而链表可以散在数据结构里面
[ , , , ,]
[ ,1, ,2,]
[4, ,3, ,]
[1|next] -> [2|next] -> [3|next] -> [4|next] -> Null

数组优点：access 快; 缺点：search、insert、delete慢
链表优点：insert、delete快；缺点：access、search慢

访问access-> O(N)
搜索search-> O(N)
插入insert-> O(1)
删除delete-> O(1)

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

if __name__ == "__main__":
    # 创建一个链表 1
    head = ListNode(1)
    head.next = ListNode(123)
    head.next.next = ListNode(456)
    print()
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
# val 和 next 会定义好

# 链表的常用操作
# 1.创建链表
# 2.添加元素
# 3.访问元素
# 4.查找元素
# 5.删除元素
# 6.链表的长度

# 1. 创建链表
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
# 2. 添加元素
new_node = ListNode(4)
new_node.next = head.next
head.next = new_node
# 3. 访问元素
current = head
while current:
    print(current.val)
    current = current.next
# 4. 查找元素
current = head
while current:
    if current.val == 2:
        print("Found:", current.val)
        break
    current = current.next
# 5. 删除元素
current = head
while current and current.next:
    if current.next.val == 3:
        current.next = current.next.next
        break
    current = current.next
# 6. 链表的长度
length = 0
current = head
while current:
    length += 1
    current = current.next
print("Length of linked list:", length) 
```

```python
from collections import deque
linkedlist = deque()

linkedlist.append(1)  # 在尾部添加元素
linkedlist.appendleft(0)  # 在头部添加元素
# 时间复杂度是O(1)，因为在尾部和头部添加元素都是常数时间操作。
print(linkedlist)

linkedlist.insert(1, 5)  # 在索引1的位置插入元素5
# **时间复杂度是O(n)**，因为在插入元素时，需要移动索引1之后的所有元素。
print(linkedlist)

element = linkedlist[1]  # 访问索引1的元素
# 时间复杂度是O(1)，因为访问索引1的元素是常数
print(element)

index = linkedlist.index(5)  # 查找元素5的索引
# 时间复杂度是O(n)，因为需要遍历整个链表来查找元素
print(index)

print(linkedlist)

linkedlist[2] = 10  # 修改索引2的元素为10
# 时间复杂度是O(1)，因为修改索引2的元素是常数
print(linkedlist)

linkedlist.remove(5)  # 删除元素5
# 时间复杂度是O(n)，因为需要遍历整个链表来查找元素
print(linkedlist)

length = len(linkedlist)  # 获取链表的长度
# 时间复杂度是O(1)，因为获取链表的长度是常数    
print(length)
```

# 12 heap 堆
大堆和小堆
大堆是把最大值放在最上面
小堆是把最小值放在最上面


```python
# 12 heapq 堆
import heapq
from heapq import heappush, heappop, heapify, nsmallest, nlargest

# 小堆
a=[3, 1, 4, 1, 5, 9, 2, 6, 5]
# heapq.heapify(a)  # 将列表a转换为最小堆
print(a)  # [1, 1, 2, 3, 5, 9, 4, 6, 5]
heapq.heappush(a, 0)  # 向堆中添加元素0
print(a)  # [0, 1, 1, 3, 5, 9, 4, 6, 5, 2]
heapq.heappop(a)  # 从堆中删除最小元素
print(a)  # [1, 1, 2, 3, 5, 9, 4, 6, 5]

# 获取堆中最小的n个元素
print(nsmallest(3, a))  # [1, 1, 2]

# 获取堆中最大的n个元素
print(nlargest(3, a))  # [9, 6, 5]

# 大堆
a = [3, 1, 4, 1, 5, 9, 2, 6, 5]
b = [-x for x in a]  # 将列表a中的元素取反
heapq.heapify(b)  # 将列表b转换为最小堆
print(b)  # [-9, -6, -5, -5, -4, -3, -2, -1, -1]

# 从大堆中获取最大的n个元素
print([-x for x in nsmallest(3, b)])  # [9, 6, 5]
# 从大堆中获取最小的n个元素
print([-x for x in nlargest(3, b)])  # [5, 4, 3]
# 最后的大堆结果
print([-x for x in b])  # [9, 6, 5, 5, 4, 3, 2, 1, 1]
```
